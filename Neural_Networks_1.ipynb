**Tensorflow** : TensorFlow acts as the powerful, low-level execution engine that handles the underlying complex mathematical computations, allows to build customizable complex deep learning models.

**Keras** : High-level interface optimized for rapid prototyping and human readability used for training and building out deep learning models

```python
import numpy as np
import pandas as pd
import matplotlib .pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Data file upload
from google.colab import files
uploaded = files.upload()
```

```python
file_name = list(uploaded.keys())[0]

''' Make a tensorflow data set

dataset = tf.data.experimental.make_csv_dataset(
    file_name,
    batch_size=32,
    label_name='Target',
    num_epochs=1,
    shuffle=True
)

print(dataset.element_spec) --> Inspect the data

'''

# Use pandas to perform feature and target split

df = pd.read_csv(file_name)
X = df.drop('Target', axis=1).to_numpy()
Y = df['Target'].to_numpy()
Y = Y.reshape(Y.shape[0],1)

print(X.shape)
print(Y.shape)

```

```python
# Normalize the data using Keras Normalisation layer

print(f"Temperature Max, Min pre normalization: {np.max(X[:,0]):0.2f}, {np.min(X[:,0]):0.2f}")
print(f"Duration Max, Min pre normalization: {np.max(X[:,1]):0.2f}, {np.min(X[:,1]):0.2f}")


normalizer = tf.keras.layers.Normalization(axis=-1)
normalizer.adapt(X)

X_norm = normalizer(X)

print(f"Temperature Max, Min pre normalization: {np.max(X_norm[:,0]):0.2f}, {np.min(X_norm[:,0]):0.2f}")
print(f"Duration    Max, Min pre normalization: {np.max(X_norm[:,1]):0.2f}, {np.min(X_norm[:,1]):0.2f}")
```

```python
# Tile/ Copy our data to increase the training set size and
# reduce the number of training epochs

X_train = np.tile(X_norm, (100,1))
Y_train = np.tile(Y, (100,1))

print(X_train.shape)
print(Y_train.shape)
```

```python
# TensorFlow Model

tf.random.set_seed(1234) # for consistent results

model = Sequential(
    [
        tf.keras.Input(shape=(2,)), # internal tensor [feature1, feature2]
        Dense(3, activation='sigmoid', name='layer_1'),
        Dense(1, activation='sigmoid', name='layer_2')
    ]
)

model.summary()
```

```python
# weights and biases tensorflow instantiated.

W1, b1 = model.get_layer("layer_1").get_weights()
W2, b2 = model.get_layer("layer_2").get_weights()
print(f"W1{W1.shape}:\n", W1, f"\nb1{b1.shape}:", b1)
print(f"W2{W2.shape}:\n", W2, f"\nb2{b2.shape}:", b2)
```

```python
model.compile(
    loss = tf.keras.losses.BinaryCrossentropy(),
    optimizer = tf.keras.optimizers.Adam(learning_rate=0.01),
)

model.fit(
    X_train,Y_train,
    epochs=10,
)
```

```python
# Checking weights once the data is trained.

W1, b1 = model.get_layer("layer_1").get_weights()
W2, b2 = model.get_layer("layer_2").get_weights()
print("W1:\n", W1, "\nb1:", b1)
print("W2:\n", W2, "\nb2:", b2)
```

```python
# Make predictions

X_test = np.array([
    [200,13.9],  # positive example
    [200,17]])   # negative example
X_test_norm = normalizer(X_test)
predictions = model.predict(X_test_norm)
print("predictions = \n", predictions)


yhat = np.zeros_like(predictions)
for i in range(len(predictions)):
    if predictions[i] >= 0.5:
        yhat[i] = 1
    else:
        yhat[i] = 0
yhat = (predictions >= 0.5).astype(int)
print(f"decisions = \n{yhat}")
```
